**Business Problem.** A customer service director wants shorter interactions to improve
productivity. An initial report associates longer handling time with lower satisfaction.
Does this justify one common target to shorten all interactions?

**Work individually, 35–40 minutes.** Use descriptive analysis to evaluate this recommendation.

**Data:** 360 synthetic requests from one quarter; one row per request. All fields are complete
and request IDs are unique. No cleaning, imputation, tests of significance or PCA are required.
Satisfaction is a complete teaching index, not a claim about real survey coverage.


### Data dictionary

| Field | Meaning / role |
|---|---|
| `request_id` | Request identifier; not a numerical measure. |
| `handling_minutes` | Active handling time, excluding queue time; numerical. |
| `queue_wait_minutes` | Waiting time before service starts; numerical. |
| `satisfaction_score` | Synthetic continuous satisfaction index, 0–100; numerical. |
| `request_complexity` | Simple, Intermediate or Complex, recorded at intake; categorical. |
| `contact_channel` | Chat, Phone or Email; categorical. |
| `resolved_first_contact` | Yes or No; categorical. |

Keep the raw values. Category order helps presentation but does not make categories continuous.
The unit is a **request**, not a customer: the file cannot assess repeated requests by the same person.
